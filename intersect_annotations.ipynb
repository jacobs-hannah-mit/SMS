> in order to run pybedtools, you need a bed file, which can be generated w/ "make bed file".

```python
#export
def make_bed_file(df, num_bp, info, no_strand):

    start=df.v_start.astype('int')
    end=df.v_end.astype('int')
    df['score']=[1]*len(df)
    df['strand']= df['strand'].str.replace('r','-')
    df['strand']= df['strand'].str.replace('f','+')
    if no_strand==True:
        forBed=pd.concat([df.chromosome, start, end, df[info], df.score, df.strand], axis=1)
        
    else:
        forBed=pd.concat([df.chromosome, start, end, df[info]], axis=1)
    return forBed 


```
